```python
import os
import argparse
from glasso_pca import GLASSO_PCA  # Ensure glasso_pca.py is in the same directory
import interactive_flow as flow   # Ensure interactive_flow.py is in the same directory

def main():
    args = parseArgs()

    gl_alpha = args.glalpha if args.glalpha is not None else -1
    pca_comp_var_threshold = args.pca_threshold if args.pca_threshold is not None else 0.1

    # Initialize GLASSO_PCA
    gp = GLASSO_PCA(
        inputfile=args.filename,
        verbose=not (args.quiet),
        gl_alpha=gl_alpha,
        pca_comp_var_threshold=pca_comp_var_threshold
    )
    gp.load()

    # Generate plots if requested
    if not args.no_plots:
        if not args.no_glasso_plots:
            gp.glasso_plots()
        if not args.no_pca_plots:
            gp.pca_plots()
        if not args.no_boxplots:
            gp.boxplots()

    # Run quick analysis if requested
    if args.quick:
        gp.quick()

    # Interactive mode
    if args.interact:
        prompt = flow.Prompt(gp)
        prompt.prompt = '> '
        prompt.cmdloop('Starting interactive console:')

def parseArgs():
    """Parses input arguments."""
    parser = argparse.ArgumentParser(description="Run GLASSO-PCA analysis.")
    parser.add_argument('filename', help='Input file to analyze')
    parser.add_argument('-quiet', action='store_true', help='Suppress stdout printouts')
    parser.add_argument('-i', '--interact', action='store_true', help='Enable interactive console')
    parser.add_argument('-quick', action='store_true', help='Run quick PCA analysis')
    parser.add_argument('-glalpha', type=float, help='Alpha parameter for GLASSO (default: auto)')
    parser.add_argument('-pca_threshold', type=float, help='Threshold for PCA explained variance (default: 0.1)')
    parser.add_argument('-no_plots', action='store_true', help='Disable all plot generation')
    parser.add_argument('-no_boxplots', action='store_true', help='Disable boxplot generation')
    parser.add_argument('-no_pca_plots', action='store_true', help='Disable PCA plot generation')
    parser.add_argument('-no_glasso_plots', action='store_true', help='Disable GLASSO plot generation')

    args = parser.parse_args()

    # Validate filename
    if not os.path.isfile(args.filename):
        print(f"Invalid filename: {args.filename}")
        raise SystemExit

    # Validate GLASSO alpha
    if args.glalpha is not None and args.glalpha < 0:
        print("GLASSO alpha must be >= 0. Defaulting to auto-selection.")
        args.glalpha = -1

    # Validate PCA threshold
    if args.pca_threshold is not None and not (0 <= args.pca_threshold <= 1):
        print("PCA threshold must be in the range [0, 1]. Defaulting to 0.1.")
        args.pca_threshold = 0.1

    return args

if __name__ == '__main__':
    main()
```
